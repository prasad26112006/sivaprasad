```python
import os
import shutil

# Create documents folder
os.makedirs("documents", exist_ok=True)

# Move all doc*.txt files into documents folder
for file in os.listdir():
    if file.startswith("doc") and file.endswith(".txt"):
        shutil.move(file, os.path.join("documents", file))

print("✅ All files moved to documents folder")
```

```python
import os
import re
from collections import defaultdict

folder_path = "documents"

inverted_index = defaultdict(set)
vocabulary = set()

for filename in os.listdir(folder_path):
    if filename.endswith(".txt"):
        with open(os.path.join(folder_path, filename), "r", encoding="utf-8") as file:
            text = file.read().lower()

            words = re.findall(r'\b[a-z]+\b', text)

            vocabulary.update(words)

            for word in set(words):
                inverted_index[word].add(filename)

print("Total Documents:", len(os.listdir(folder_path)))
print("Vocabulary Size:", len(vocabulary))

print("\nSample Inverted Index:\n")

count = 0
for word in sorted(inverted_index):
    print(f"{word} -> {sorted(inverted_index[word])}")
    count += 1
    if count == 30:
        break
```
